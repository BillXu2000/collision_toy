```python
import sympy
sympy.init_printing()

l0 = sympy.symbols('l_0')
k = sympy.symbols('k')
n = 3

def norm(x):
    ans = 0
    for i in x:
        ans += i**2
    return sympy.sqrt(ans)

def dist(x):
    return norm(x) - l0

def energy(x):
    return k * dist(x)**2 / 2
```

```python
xs = [sympy.symbols(f'x_{i}') for i in range(n)]
df = 0
for i, x in enumerate(xs):
    df += sympy.symbols(f'dx_{i}') * sympy.diff(energy(xs), x)
df = sympy.simplify(df)

df
```

```python
ddf = 0
for i, x in enumerate(xs):
    ddf += sympy.symbols(f'dx_{i}') * sympy.diff(df, x)
ddf = sympy.simplify(ddf)
# ddf = sympy.collect(ddf, sympy.symbols('dx_0'))
# ddf = sympy.aprt(ddf)

ddf
```

```python
n = norm(xs)
dn = 0

def ddx(f) :
    df = 0
    for i, x in enumerate(xs):
        df += sympy.Symbol('(\partial x_{%d})' % i) * sympy.diff(f, x)
    return df
ans = ddx(ddx(n))
ans = sympy.simplify(ans)
ans
```

```python
n = norm(xs)

def d2dx2(f):
    m = []
    for i, x in enumerate(xs):
        df = sympy.diff(f, x)
        ddf = 0
        for j, y in enumerate(xs):
            ddf += sympy.Symbol('(\partial x_{%d})' % j) * sympy.diff(df, y)
        m.append(ddf)
    return sympy.Matrix(m).transpose()
ans = d2dx2(n)
ans = sympy.simplify(ans)
# ans
# sympy.srepr(ans[0])
ans[0]
```
